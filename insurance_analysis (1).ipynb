```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('insurance.csv')
print(df.head())
```

```python
df.describe(include='all')
df.isnull().sum()
```

```python
# 3. Exploratory Data Analysis (EDA)
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Distribution of Charges
sns.histplot(df['charges'], kde=True, ax=axes[0,0], color='blue')
axes[0,0].set_title('Distribution of Charges')

# Charges by Smoker Status
sns.boxplot(x='smoker', y='charges', data=df, ax=axes[0,1])
axes[0,1].set_title('Insurance Charges vs Smoker Status')

# BMI vs Charges
sns.scatterplot(x='bmi', y='charges', hue='smoker', data=df, ax=axes[1,0])
axes[1,0].set_title('BMI vs Charges')

# Age vs Charges
sns.scatterplot(x='age', y='charges', hue='smoker', data=df, ax=axes[1,1])
axes[1,1].set_title('Age vs Charges')

plt.tight_layout()
plt.show()
```

```python
# 4. Correlation Analysis
# One-hot encode categorical variables to check global correlation
df_encoded = pd.get_dummies(df, drop_first=True)
plt.figure(figsize=(10, 6))
sns.heatmap(df_encoded.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix')
plt.show()
```
